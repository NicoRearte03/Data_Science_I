### Hallazgos Estadísticos y Anomalías Detectadas:

1. **Sesgo y presencia de outliers en la distribución de montos:**  
   Al analizar los valores del `describe()`, se observa una marcada diferencia entre la mediana (percentil 50%) y el valor máximo en las variables de montos (`amount` / `amount_ratio`). Esto indica un sesgo positivo (hacia la derecha) y la presencia de *outliers* significativos, lo cual es característico de operaciones financieras atípicas o posibles eventos de fraude.

2. **Marcado desbalance en la variable objetivo (`is_fraud`):**  
   La media de la columna target se sitúa cercana a 0.017 (1.71%), lo que confirma que las transacciones fraudulentas constituyen una minoría absoluta en el conjunto de datos. Este desbalance refleja fielmente un entorno de producción real, pero requerirá métricas específicas de evaluación más allá del simple *accuracy*.

```python
# =========================================================
# BLOQUE 3: DIAGNÓSTICO DE DATOS FALTANTES
# =========================================================

# 1. Calcular el porcentaje exacto de nulos por columna
porcentaje_nulos = (df.isnull().sum() / len(df)) * 100

# 2. Ordenar de mayor a menor
porcentaje_nulos_ordenado = porcentaje_nulos.sort_values(ascending=False)

# 3. Crear un DataFrame empaquetado para mostrarlo prolijo en pantalla
df_diagnostico_nulos = pd.DataFrame({
    'Columna': porcentaje_nulos_ordenado.index,
    '% Datos Nulos': porcentaje_nulos_ordenado.values
})

# Mostrar la tabla formateada
display(df_diagnostico_nulos)
```

```python
# =========================================================
# BLOQUE 4: LÓGICA ALGORÍTMICA Y FUNCIONES PERSONALIZADAS
# =========================================================

# 1. FUNCIÓN FORMAL (def)
# Evaluamos la columna 'amount_vs_avg_ratio' para categorizar el nivel de sospecha según el desvío del gasto
def clasificar_nivel_riesgo(ratio):
    """
    Clasifica el riesgo de una transacción comparando el monto contra
    el promedio histórico del usuario.
    """
    if ratio > 5.0:
        return 'Alto Riesgo'
    elif ratio > 2.0:
        return 'Riesgo Medio'
    else:
        return 'Riesgo Bajo'

# Aplicamos la función formal con .apply() para crear una nueva columna
df['nivel_riesgo_monto'] = df['amount_vs_avg_ratio'].apply(clasificar_nivel_riesgo)


# 2. FUNCIÓN EXPRESS (lambda)
# Convertimos el monto original ('amount') de unidades a miles para simplificar la lectura
df['amount_en_miles'] = df['amount'].apply(lambda x: round(x / 1000, 2))


# Muestra de verificación con las columnas originales y las dos nuevas generadas
display(df[['amount', 'amount_vs_avg_ratio', 'nivel_riesgo_monto', 'amount_en_miles']].head(10))
```
